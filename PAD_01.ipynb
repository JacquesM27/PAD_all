```python
import numpy as np
import pandas as pd
```

```python
print('Task 1')

# Wczytanie danych
data = pd.read_csv('data/01/Zadanie_1.csv', delimiter=';')
matrix = data.to_numpy()

# a) Liczba komórek w macierzy
print('Dane podstawowe bez obróbki')
print('\tLiczba komórek:  ', matrix.size)

# b) Liczba wierszy i kolumn
print('\tLiczba wierszy:  ', matrix.shape[0])
print('\tLiczba kolumn:   ', matrix.shape[1])

# c) Statystyki opisowe dla pełnych danych (z NaN w macierzy)
print('\tŚrednia:         ', np.mean(matrix))
print('\tMediana:         ', np.median(matrix))
print('\tWariancja:       ', np.var(matrix))

# d) Usunięcie braków danych - maskowanie NaN
# Oba zapisy dają taki sam wynik ale ten drugi był na wykładach
# mask = ~np.isnan(matrix)
# masked_matrix = matrix[mask]
# print('Pominięcie brakujących danych (usunięcie NaN)')
# print('\tŚrednia:         ', np.mean(masked_matrix))
# print('\tMediana:         ', np.median(masked_matrix))
# print('\tWariancja:       ', np.var(masked_matrix))

masked_matrix = np.ma.masked_invalid(matrix)
print(masked_matrix)
print('Pominięcie brakujących danych (maskowanie NaN)')
print('\tŚrednia:         ', np.ma.mean(masked_matrix))
print('\tMediana:         ', np.ma.median(masked_matrix))
print('\tWariancja:       ', np.ma.var(masked_matrix))
```

```python
print('Task 2')
data = np.loadtxt('data/01/Zadanie_2.csv', delimiter=';')
eigenvalues, eigenvectors = np.linalg.eig(data)

print('\n================== Wartości własne ==================\n', eigenvalues)
print('\n================== Wektory własne ===================\n', eigenvectors)
print('\n================== Macierz odwrotna ================\n', np.linalg.inv(data))
```

```python
print('Task 3')

A = np.loadtxt('data/01/Zadanie_3_macierz_A.csv', delimiter=',')
B = np.loadtxt('data/01/Zadanie_3_macierz_B.csv', delimiter=',')

def csm(a, b):
    num=np.dot(a, b.T)
    p1= np.sqrt(np.sum(a ** 2, axis=1))[:, np.newaxis]
    p2= np.sqrt(np.sum(b ** 2, axis=1))[np.newaxis, :]
    return num/(p1*p2)

similarity = csm(A,B)
print(similarity)
```
