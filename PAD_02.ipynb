```python
import numpy as np
import pandas as pd
```

```python
print('Task 1')
data_1 = pd.read_csv('data/02/president_heights.csv')
heights = np.array(data_1['height(cm)'])
print(heights)

mean_height = heights.mean()
standard_deviation = heights.std()
minimum_height = heights.min()
maximum_height = heights.max()

print("Mean height:       ", mean_height)
print("Standard deviation:", standard_deviation)
print("Minimum height:    ", minimum_height)
print("Maximum height:    ", maximum_height)

percentile_25 = np.percentile(heights, 25)
median = np.median(heights)
percentile_75 = np.percentile(heights, 75)

print("25th percentile:   ", percentile_25)
print("Median:            ", median)
print("75th percentile:   ", percentile_75)
```

```python
print('Task 2')
data_2 = pd.read_csv('data/02/Zadanie_2 1.csv', delimiter=';', header=None)
values = data_2.values
eig_values, eig_vectors = np.linalg.eig(values)
print("Wartości własne:   ", eig_values)
print("Wektory własne:    ", eig_vectors)
print("Macierz odwrotna:  ", np.linalg.inv(values))
```

```python
print('Task 3')
rainfall = pd.read_csv('data/02/Seattle2014.csv')['PRCP'].values
inches = rainfall / 254.0  # 1/10mm -> inches
# inches.shape

%matplotlib inline
import matplotlib.pyplot as plt
import seaborn; seaborn.set_theme()  # set plot styles
plt.hist(inches, 40)
plt.xlabel("Opady (cale)")
plt.ylabel("Lliczba dni")
plt.title('Histogram opadów dla Seattle (2014)')
plt.show()

days_without_rain = (inches == 0).sum()
days_with_any_rain = (inches > 0).sum()
days_with_more_than_0_5_inches = (inches > 0.5).sum()
rainy_days_with_less_than_0_2_inches = ((inches < 0.2) & (inches > 0)).sum()

print("Number of days without rain:      ", days_without_rain)
print("Number of days with rain:         ", days_with_any_rain)
print("Days with more than 0.5 inches:   ", days_with_more_than_0_5_inches)
print("Rainy days with < 0.2 inches:     ", rainy_days_with_less_than_0_2_inches)

data_3 = pd.read_csv('data/02/Seattle2014.csv')
data_3['PRCP_INCHES'] = data_3['PRCP'] / 254.0
rainy_days = data_3[data_3['PRCP_INCHES'] > 0]
median_rainy = rainy_days['PRCP_INCHES'].median()

summer_days = data_3[(data_3['DATE'] >= 20140621) & (data_3['DATE'] <= 20140922)]
median_summer = summer_days['PRCP_INCHES'].median()
max_summer = summer_days['PRCP_INCHES'].max()

non_summer_days = data_3[(data_3['DATE'] < 20140621) | (data_3['DATE'] > 20140922)]
max_non_summer = non_summer_days['PRCP_INCHES'].max()

print("Median precip on rainy days in 2014 (inches):   ", median_rainy)
print("Median precip on summer days in 2014 (inches):  ", median_summer)
print("Maximum precip on summer days in 2014 (inches): ", max_summer)
print("Maximum precip on non-summer rainy days (inches):", max_non_summer)
```
