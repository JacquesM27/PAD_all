Wgraj zawartość pliku orders.csv do ramki danych.
Przyjrzyj się danym, wykorzystując poznane metody (describe, info, head) zanalizuj dane.

Następnie:



1. Zamień order_date na format daty. Upewnij się, że w rezultacie data będzie w formacie: rok-miesiąc-dzień;
2. Sprawdź ile i jakie unikalne wartości znajdują się w kolumnie T-shirt category;
3. Wyczyść kolumnę tshirt_category normalizując nazwy produktów. Upewnij się, że dla każdej kategorii funkcjonuje jedna nazwa, np. zamiast white i wh jest tylko jedna wersja white.
4. Rozdziel kolumnę tshirt category na 3 kolumny: tshirt_type, która będzie mówiła o tym czy jest to koszulka, bluza czy coś innego, tshirt_department, która będzie informowała o tym czy jest męska czy żeńska, i tshirt_colour, która będzie informowała o kolorze. Skorzystaj z metody apply.
5. Znajdź zamówienia złożone między 31 grudnia 2014 a 2 sierpnia 2016.


```python
import pandas as pd

orders = pd.read_csv("data/04/orders.csv", delimiter=",")
print("============= describe =============")
print(orders.describe())
print("============= info =============")
print(orders.info())
print("============= head =============")
print(orders.head())

orders["order_date"] = pd.to_datetime(orders["order_date"])
print("============= to_datetime =============")
print(orders.head())
print("============= unique T-shirt category =============")
print(orders["tshirt_category"].unique())
print("Unique T-shirt categories:", orders["tshirt_category"].unique().size)

orders["tshirt_category"] = orders["tshirt_category"].replace({
    "Wh Tshirt F": "White T-Shirt F",
    "Wh Tshirt M": "White T-Shirt M",
    "Bl Tshirt F": "Black T-Shirt F",
    "Bl Tshirt M": "Black T-Shirt M"
})

print("============= normalized unique T-shirt category =============")
print(orders["tshirt_category"].unique())
print("Unique T-shirt categories:", orders["tshirt_category"].unique().size)

def split_tshirt_category(category):
    if "Hoodie" in category:
        tshirt_type = "Hoodie"
    elif "Shirt" in category:
        tshirt_type = "Shirt"
    else:
        tshirt_type = "Other"
    
    if "M" in category:
        tshirt_department = "Man"
    elif "F" in category:
        tshirt_department = "Woman"
    else:
        tshirt_department = "Unisex"
        
    if "White" in category:
        tshirt_colour = "White"
    elif "Black" in category:
        tshirt_colour = "Black"
    else:
        tshirt_colour = "Unknown"
    
    return pd.Series([tshirt_type, tshirt_department, tshirt_colour])

orders[["tshirt_type", "tshirt_department", "tshirt_colour"]] = orders["tshirt_category"].apply(split_tshirt_category)
print("============= split T-shirt category =============")
print(orders.head())
print("============= orders from 31.12.2014 to 2.08.2016 =============")
# wynik tego i tego poniżej jest taki sam
# start_date = '2014-12-31'
# end_date = '2016-08-02'
# filtered_orders = orders[(orders['order_date'] > start_date) & (orders['order_date'] < end_date)]
# print(filtered_orders)
print(orders.query("'2014-12-31' < order_date < '2016-08-02'"))
```
